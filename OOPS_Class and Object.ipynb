```python
from abc import ABC, abstractmethod

# Parent Class
class Parent_class(ABC):

    @abstractmethod
    def method_name(self):
        pass


# Child Class
class Child_class(Parent_class):

    def method_name(self):
        print("Abstract method implemented in child class")


# Object Creation
obj = Child_class()
obj.method_name()
```

```python
class Temperature:   # blueprint

    def __init__(self):
        self.__temp = 0   # Privet variable.

    def set_temp(self, temp):   # setter method used to update temp
        if -50 <= temp <= 60:    # Validation logic means temp >= -50
                                  # and temp <= 60
            self.__temp = temp
            
        else:
            
            print("Invalid Temperature")

    def get_temp(self): 
                                # getter method, return the privet value. 
        return self.__temp

t = Temperature()

t.set_temp(30)   # here we put value more than 60 then else condition apply.

print("Temperature:", t.get_temp())
        
    
```

```python
from abc import ABC, abstractmethod

# Abstract class
class Vehicle(ABC):


    def __init__(self, brand):
        self.brand = brand

    @abstractmethod
    def start(self):
            pass

# Child class
class Car(Vehicle):

    def __init__(self, brand):
        super().__init__(brand)   # calling parent constructor

    def start(self):
        print(f"{self.brand} car is starting...")


# object creation
car1 = Car("TATA")

print("Brand:", car1.brand)
car1.start()
```

```python
# employee management system
# parent class :
# attributes : maneger_name , salary
# show_employee_details()
# abstract method : project_name
# child class : attribute: manager_name,emp_id,emp_name,emp_salary
# implementation of abstract
# display all details

```

```python
from abc import ABC, abstractmethod

class Manager(ABC):

    def __init__(self,manager_name,manager_salary):
        self.manager_name = manager_name
        self.manager_salary = manager_salary

    def show_employee_details(self):
        print("Manager Name:", self.manager_name)
        print("Manager Salary:", self.manager_salary)

    @abstractmethod
    def project_name(self):
        pass

# child class
class Employee(Manager):

    def __init__(self,manager_name, salary, emp_id, emp_name, emp_salary):

        # calling parent cnstructor

        super().__init__(manager_name, salary)
        
        self.emp_id = emp_id
        self.emp_name = emp_name
        self.emp_salary = emp_salary

    def project_name(self):
        print("Project name : Employee Management System")

        # Display all details

    def display_all_details(self):
        
        
        self.show_manager_details()   # parent method

        print("Employee ID:", self.emp_id)
        print("Employee Name:", self.emp_name)
        print("Employee Salary:", self.emp_salary)


    
        self.project_name()

# Object creatiion
emp1 = Employee("AZIM", 100000, 1525, " Ramya",10000,)
emp1.display_all_details()
        

    
```
